```python
#Media ( 2020 )
df_2020 = dataframe[dataframe['Year'] == 2020]
poblacion_2020 = df_2020['Annual change in population']

#identificar minimos y maximos

entidad_min = df_2020.loc[poblacion_2020.idxmin()]["Entity"]
entidad_max = df_2020.loc[poblacion_2020.idxmax()]["Entity"]

print ("Media: ", poblacion_2020.mean() )
print("Mediana :",poblacion_2020.median())
print("Varianza :",poblacion_2020.var())
print("Desviacion estandar :",poblacion_2020.std())
print("Minimo :" , poblacion_2020.min() , "en" , entidad_min)
print("Maximo :",poblacion_2020.max() , "en" , entidad_max)
```
