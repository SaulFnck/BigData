```python
import pandas as pd
import requests

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
#REnombrar columnas
df.rename(columns={"entity" : "pais"},inplace=True) #Implace para mantener el cambio
df.head()
```

```python
# case ayuda a diferenciar mayusculas con minusculas ( estricto )
# na ayuda a controlar nulos
# regex permite buscar con simbolos:
#   ^ = Que empiece con
#   $ = que termine con
df[(df["year"] == 1970) & (df["pais"].str.contains("^K",case=False,regex=True)) ][["pais","year","life_expectancy_0"]]
```

```python
df.set_index("pais")
```

```python
df.loc[df["code"].notnull(),["year","life_expectancy_0"]].head()
```

```python
#Borra valores nulos
df.dropna(subset=["code"],inplace=True)
df.isnull().sum()
```

```python
#agrupar
resultado = (
    df.groupby("pais")["life_expectancy_0"].mean().sort_values(ascending=True)
)

resultado.head(10)
```

```python
new_df = df[df["year"].between(1970,2023)]
resultado = (
    new_df.groupby("pais")["life_expectancy_0"].mean().sort_values(ascending=True)
)
print(resultado)
```
